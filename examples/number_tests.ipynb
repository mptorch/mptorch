```python
from mptorch.number import FloatingPoint, SubnormalsMode
from mptorch.quant import float_quantize

fp8e5m2 = FloatingPoint(exp=5, man=2, subnormals=SubnormalsMode.SUBNORMALS)

fp8m5m2_quant = lambda x: float_quantize(x, exp=5, man=2, rounding='RNE', subnormals=True, saturate=False)
```

```python
print("FloatingPoint Statistics:")
print("max_normal:    ", fp8e5m2.normal_max)
print("min_normal:    ", fp8e5m2.normal_min)
print("max_subnormal: ", fp8e5m2.subnormal_max)
print("min_subnormal: ", fp8e5m2.subnormal_min)
```

```python
import torch
x = torch.tensor([200000.0])
print(fp8m5m2_quant(x))
```
